```python
from metadrive.envs import MetaDriveEnv
from IPython.display import Image

import cv2
env = MetaDriveEnv(dict(log_level=50))
env.reset()
for i in range(100):
    env.step([0,0])
    if i>=0:
        env.render(mode="topdown",
                   window=False,
                   screen_size=(400, 200),
                   screen_record=True,
                   text={"Step": i})
env.top_down_renderer.generate_gif(gif_name="top-down.gif")
env.reset()

env.close()

Image(open("top-down.gif", 'rb').read())
```
